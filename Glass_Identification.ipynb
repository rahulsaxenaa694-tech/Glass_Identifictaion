```python
print(glass["Type"].value_counts().sort_index())
```

```python
columns_to_check = ["RI", "Na", "Mg", "Al"]

for column in columns_to_check:
    values = glass[column].to_numpy()

    average = np.mean(values)
    smallest = np.min(values)
    largest = np.max(values)
    value_range = largest - smallest

    print(column)
    print("Average:", average)
    print("Smallest:", smallest)
    print("Largest:", largest)
    print("Range:", value_range)
    print()
```

```python
for glass_type in range(1, 8):
    mg_values = glass[glass["Type"] == glass_type]["Mg"]

    if len(mg_values) > 0:
        print("Type", glass_type, "average Mg:", np.mean(mg_values))
```

```python
top_10_sodium = glass.sort_values(by="Na",ascending=False).head(10)

print(top_10_sodium[["Id", "Na", "Type"]])
```

```python
print("Missing values in each column:")
print(glass.isnull().sum())

print("Number of duplicate rows:")
print(glass.duplicated().sum())
```

```python
chemicals = ["RI", "Na", "Mg", "Al", "Si", "K", "Ca", "Ba", "Fe"]

for glass_type in range(1, 8):
    type_samples = glass[glass["Type"] == glass_type]

    if len(type_samples) > 0:
        print("\nGlass type:", glass_type)
        print(type_samples[chemicals].mean())
```
